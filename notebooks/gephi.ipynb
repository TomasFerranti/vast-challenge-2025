```python
import json
import networkx as nx
import numpy as np
import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import os

from math import radians, sin, cos, sqrt, atan2
```

```python
def load_json(file_path):
    with open(file_path, 'r') as f:
        return json.load(f)
```

```python
root = './2025-data-main/MC2_release/'

filah_data = load_json(root + 'FILAH.json')
trout_data = load_json(root + 'TROUT.json')
journalist_data = load_json(root + 'journalist.json')
road_map_data = load_json(root + 'road_map.json')
```

```python
def create_gephi_files(data, output_dir='gephi_output'):
    """
    Transform data dictionary with 'nodes' and 'links' into Gephi input format.
    Creates two files: nodes.csv and edges.csv in the specified output directory.
    """
    # Create output directory if it doesn't exist
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    # Process nodes
    nodes_df = pd.DataFrame(data['nodes'])
    
    # Add required Gephi node columns if they don't exist
    if 'Id' not in nodes_df.columns:
        nodes_df['Id'] = nodes_df['id']
    if 'Label' not in nodes_df.columns:
        nodes_df['Label'] = nodes_df['label']
    
    # Process edges
    edges_df = pd.DataFrame(data['links'])
    
    # Add required Gephi edge columns if they don't exist
    if 'Source' not in edges_df.columns:
        edges_df['Source'] = edges_df['source']
    if 'Target' not in edges_df.columns:
        edges_df['Target'] = edges_df['target']
    if 'Type' not in edges_df.columns:
        edges_df['Type'] = 'Directed'  # Default to directed edges
    
    # Save to CSV files
    nodes_df.to_csv(f'{output_dir}/nodes.csv', index=False)
    edges_df.to_csv(f'{output_dir}/edges.csv', index=False)
    
    print(f"Gephi files created in {output_dir}/")
    print(f"Nodes file: {len(nodes_df)} nodes")
    print(f"Edges file: {len(edges_df)} edges")
```

```python
# Example usage:
create_gephi_files(journalist_data)
```
